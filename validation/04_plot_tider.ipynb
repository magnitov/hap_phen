```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
allele_rates = [[55.1, 44.9], [9.3, 90.7]]
allele_rates = pd.DataFrame(allele_rates)
allele_rates.columns = ['G', 'T']
allele_rates.index = ['WT', 'Edited']

T_convertion = allele_rates.loc['Edited', 'T'] - allele_rates.loc['WT', 'T']
T_max = allele_rates.loc['WT', 'G']
editing_efficiency = round(T_convertion / T_max * 100, 1)
print(f'HDR efficiency: {editing_efficiency}%')

allele_rates
```

```python
tider = [[-10, 0.00],
         [-9, 0.00],
         [-8, 0.00],
         [-7, 0.00],
         [-6, 0.00],
         [-5, 0.00],
         [-4, 0.00],
         [-3, 0.20],
         [-2, 0.00],
         [-1, 1.20],
         [0, 3.40],
         [1, 0.90],
         [2, 0.00],
         [3, 0.00],
         [4, 0.00],
         [5, 1.60]]

tider = pd.DataFrame(tider)
tider.columns = ['pos', 'fraction']
tider
```

```python
plt.figure(figsize = (6, 3))
plt.title('Indel spectrum')
plt.bar(tider['pos'], tider['fraction'], edgecolor = 'k', color = '#CCCCCC')
plt.xlabel('Position from cut site')
plt.ylabel('Percentage of edits')
plt.xticks(tider['pos'])
plt.xlim((-10.5, 5.5))
plt.ylim((0, 4))
plt.tight_layout()
plt.savefig('/DATA/users/m.magnitov/hap_phen/figures/tider_quantification.pdf', bbox_inches = 'tight')
plt.show()
```
